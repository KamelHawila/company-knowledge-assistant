Clean documents before retrieval

```python
import re
from dataclasses import replace


def clean_text(text: str) -> str:
    """Normalize document text for chunking, embeddings, and search."""
    text = text.strip()
    text = text.replace("\r\n", "\n").replace("\r", "\n")
    text = re.sub(r"[ \t]+", " ", text)
    text = re.sub(r"\n{3,}", "\n\n", text)
    text = re.sub(r" *\n *", "\n", text)
    return text


clean_docs: List[CompanyDoc] = [
    replace(doc, text=clean_text(doc.text))
    for doc in docs
]

clean_docs[0]
```

Chunk documents

```python
def chunk_text(text: str, chunk_size: int = 80, chunk_overlap: int = 16) -> List[str]:
    """Split text into overlapping word chunks."""
    if chunk_size <= 0:
        raise ValueError("chunk_size must be greater than 0")
    if chunk_overlap < 0:
        raise ValueError("chunk_overlap cannot be negative")
    if chunk_overlap >= chunk_size:
        raise ValueError("chunk_overlap must be smaller than chunk_size")

    words = text.split()
    step = chunk_size - chunk_overlap
    chunks = []

    for start in range(0, len(words), step):
        end = start + chunk_size
        chunk = " ".join(words[start:end])
        chunks.append(chunk)

        if end >= len(words):
            break

    return chunks


def chunk_documents(documents: List[CompanyDoc], chunk_size: int = 80, chunk_overlap: int = 15) -> list[dict]:
    """Create searchable chunks while keeping each chunk linked to its source document."""
    chunked_docs = []

    for doc in documents:
        text_chunks = chunk_text(doc.text, chunk_size=chunk_size, chunk_overlap=chunk_overlap)

        for chunk_index, chunk in enumerate(text_chunks):
            chunked_docs.append({
                "chunk_id": f"{doc.doc_id}-CHUNK-{chunk_index + 1:03d}",
                "doc_id": doc.doc_id,
                "category": doc.category,
                "title": doc.title,
                "source": doc.source,
                "chunk_index": chunk_index,
                "text": chunk,
                "word_count": len(chunk.split()),
            })

    return chunked_docs


chunks = chunk_documents(clean_docs, chunk_size=80, chunk_overlap=16)

print(f"Created {len(chunks)} chunks from {len(clean_docs)} documents")
chunks[0]
```

Experiment with chunk sizes

```python
chunk_size_options = [40, 80, 120]
experiment_results = []

for size in chunk_size_options:
    overlap = max(5, size // 5)
    candidate_chunks = chunk_documents(clean_docs, chunk_size=size, chunk_overlap=overlap)
    word_counts = [chunk["word_count"] for chunk in candidate_chunks]

    experiment_results.append({
        "chunk_size": size,
        "chunk_overlap": overlap,
        "total_chunks": len(candidate_chunks),
        "average_words_per_chunk": round(sum(word_counts) / len(word_counts), 1),
        "smallest_chunk_words": min(word_counts),
        "largest_chunk_words": max(word_counts),
    })

experiment_results
```

### Chunk Size Decision

I tested chunk sizes of 40, 80, and 120 words.

A chunk size of 40 created too many small chunks, which may improve precision but risks losing context.

A chunk size of 120 produced fewer chunks, but for these short policy documents it did not provide a major advantage.

I selected a chunk size of 80 words with 16 words overlap because it gives a good balance between context preservation and retrieval precision.

Store embeddings

```python
import numpy as np
from sentence_transformers import SentenceTransformer


embedding_model = SentenceTransformer("all-MiniLM-L6-v2")

chunk_texts = [chunk["text"] for chunk in chunks]

embedding_matrix = embedding_model.encode(
    chunk_texts,
    convert_to_numpy=True,
    normalize_embeddings=True,
)

chunk_metadata = [
    {key: value for key, value in chunk.items() if key != "text"}
    for chunk in chunks
]

print(f"Stored {embedding_matrix.shape[0]} chunk embeddings")
print(f"Each embedding has {embedding_matrix.shape[1]} dimensions")
print(f"Embedding matrix shape: {embedding_matrix.shape}")
```

```python
embedding_store = {
    "model_name": "all-MiniLM-L6-v2",
    "embeddings": embedding_matrix,
    "texts": chunk_texts,
    "metadata": chunk_metadata,
}

print(embedding_store["metadata"][0])
print(embedding_store["embeddings"][0][:10])
```

Implement similarity search

```python
def retrieve(query: str, embedding_store: dict, top_k: int = 3) -> list[dict]:
    """Return the most similar chunks for a user query."""
    query_embedding = embedding_model.encode(
        [query],
        convert_to_numpy=True,
        normalize_embeddings=True,
    )[0]

    similarities = embedding_store["embeddings"] @ query_embedding
    top_indices = np.argsort(similarities)[::-1][:top_k]

    results = []
    for rank, index in enumerate(top_indices, start=1):
        metadata = embedding_store["metadata"][index]
        results.append({
            "rank": rank,
            "score": float(similarities[index]),
            "chunk_id": metadata["chunk_id"],
            "doc_id": metadata["doc_id"],
            "category": metadata["category"],
            "title": metadata["title"],
            "source": metadata["source"],
            "text": embedding_store["texts"][index],
        })

    return results


results = retrieve("How many vacation days do employees get?", embedding_store, top_k=3)
results
```

```python
for result in results:
    print(f"Rank {result['rank']} | Score: {result['score']:.3f} | {result['chunk_id']} | {result['title']}")
    print(result["text"][:300])
    print("-" * 80)
```

Integrate ChromaDB

```python
from pathlib import Path
import chromadb


project_root = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
chroma_path = project_root / "data" / "chroma_company_kb"

chroma_client = chromadb.PersistentClient(path=str(chroma_path))

collection_name = "company_knowledge_chunks"

existing_collection_names = [collection.name for collection in chroma_client.list_collections()]
if collection_name in existing_collection_names:
    chroma_client.delete_collection(collection_name)

collection = chroma_client.create_collection(
    name=collection_name,
    metadata={"hnsw:space": "cosine"},
)

collection.add(
    ids=[metadata["chunk_id"] for metadata in chunk_metadata],
    documents=chunk_texts,
    embeddings=embedding_matrix.tolist(),
    metadatas=chunk_metadata,
)

print(f"Stored {collection.count()} chunks in ChromaDB")
print(f"ChromaDB path: {chroma_path}")
```

Re-implement search using ChromaDB

```python
def retrieve_chroma(query: str, collection, top_k: int = 3) -> list[dict]:
    """Return the most similar chunks using ChromaDB."""
    query_embedding = embedding_model.encode(
        [query],
        convert_to_numpy=True,
        normalize_embeddings=True,
    )[0]

    response = collection.query(
        query_embeddings=[query_embedding.tolist()],
        n_results=top_k,
        include=["documents", "metadatas", "distances"],
    )

    results = []
    for rank, (chunk_id, document, metadata, distance) in enumerate(
        zip(
            response["ids"][0],
            response["documents"][0],
            response["metadatas"][0],
            response["distances"][0],
        ),
        start=1,
    ):
        results.append({
            "rank": rank,
            "score": 1 - float(distance),
            "distance": float(distance),
            "chunk_id": chunk_id,
            "doc_id": metadata["doc_id"],
            "category": metadata["category"],
            "title": metadata["title"],
            "source": metadata["source"],
            "text": document,
        })

    return results


chroma_results = retrieve_chroma("How many vacation days do employees get?", collection, top_k=3)
chroma_results
```

```python
for result in chroma_results:
    print(f"Rank {result['rank']} | Score: {result['score']:.3f} | {result['chunk_id']} | {result['title']}")
    print(result["text"][:300])
    print("-" * 80)
```

Compare speed vs manual search

```python
import time


test_queries = [
    "How many vacation days do employees get?",
    "When do VPN sessions disconnect?",
    "What happens if an invoice is unpaid?",
    "Can employees use public Wi-Fi?",
]


def average_search_time(search_function, runs: int = 5) -> float:
    start = time.perf_counter()
    for _ in range(runs):
        for query in test_queries:
            search_function(query)
    end = time.perf_counter()
    return (end - start) / (runs * len(test_queries))


manual_avg_seconds = average_search_time(lambda query: retrieve(query, embedding_store, top_k=3))
chroma_avg_seconds = average_search_time(lambda query: retrieve_chroma(query, collection, top_k=3))

speed_comparison = {
    "manual_search_avg_ms": round(manual_avg_seconds * 1000, 2),
    "chroma_search_avg_ms": round(chroma_avg_seconds * 1000, 2),
}

speed_comparison
```

### ChromaDB Decision

Manual search is useful for learning because it shows the math directly: embed the query, compute similarity scores, sort the chunks.

ChromaDB stores the same chunks, embeddings, and metadata in a persistent vector database. This becomes more useful as the document collection grows, because the database can manage indexing, storage, and retrieval instead of keeping everything in a plain Python dictionary.
